```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import xgboost as xgb
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
df=pd.read_csv('Merge_data_gaussian-v1.csv')
```

```python
df['move_type'] = df['total_count'].apply(lambda x: 0 if x == 0 else 1 if 1 <= x <= 4 else 2)
```

```python
del df['Unnamed: 0']
```

```python
import seaborn as sns
correlation_matrix = df.corr()

# Create a heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```

```python
drop_features = ['total_count','move_type']
target_feature = 'move_type'
```

```python
X = df.drop(drop_features, axis=1)
y = df[target_feature]
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.metrics import accuracy_score
scores=[]
for i in range(1000):
    X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=i)
    model = LogisticRegression(max_iter=1000,multi_class='ovr', random_state=i,C=10,penalty='l2',solver='newton-cg')
    model.fit(X_train,y_train)
    y_pred=model.predict(X_test)
    scores.append(accuracy_score(y_test, y_pred))
```

```python
np.argmax(scores)
```

```python
scores[np.argmax(scores)]
```

```python
from sklearn.metrics import classification_report
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=690)
```

```python
model = LogisticRegression(max_iter=1000,multi_class='ovr', random_state=690,C=1,penalty='l1',solver='liblinear')
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)

print(f"Accuracy: {accuracy:.2f}")
print("Classification Report:\n", report)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
logreg_classifier = LogisticRegression(max_iter=100, random_state=42,multi_class='ovr')

# Define a grid of hyperparameters to search
param_grid = {
    'penalty': ['l1', 'l2', 'elasticnet'],
    'C': [0.001, 0.01, 0.1, 1, 10, 100],
    'solver': ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga'],
}

# Create the GridSearchCV object
grid_search = GridSearchCV(estimator=logreg_classifier, param_grid=param_grid, scoring='accuracy', cv=5)

# Perform the grid search on the training data
grid_search.fit(X_train, y_train)

# Get the best parameters and the best model
best_params = grid_search.best_params_
best_model = grid_search.best_estimator_

# Make predictions on the test set using the best model
y_pred = best_model.predict(X_test)

# Evaluate the performance of the best model
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)

print("Best Parameters:", best_params)
print(f"Accuracy with Best Model: {accuracy:.2f}")
print("Classification Report:\n", report)
```
