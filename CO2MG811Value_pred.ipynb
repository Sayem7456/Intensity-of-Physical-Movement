```python
import pandas as pd
```

```python
df=pd.read_csv("database_gas.csv")
```

```python
df.describe()
```

```python
drop=['CO2MG811Value','timestamp']
target_feature = 'CO2MG811Value'
```

```python
X = df.drop(drop, axis=1)
X = X.values
y = df[target_feature].values
```

```python
X.shape
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=2,shuffle=True)
```

```python
from xgboost import XGBRegressor
my_model = XGBRegressor(n_estimators=300, learning_rate=0.5,max_depth=5,eval_metric='mae')
```

```python
my_model.fit(X_train, y_train)
```

```python
my_model.score(X_test,y_test)
```

```python
my_model.score(X_train,y_train)
```

```python
y_pred = my_model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(y_test, y_pred)
```

```python
my_model.fit(X,y)
```

```python
my_model.score(X,y)
```

```python
y_pred1 = my_model.predict(X)
```

```python
mean_squared_error(y, y_pred1)
```

```python
y_pred1=pd.DataFrame(y_pred1)
y_pred1.reset_index(inplace=True)
y_pred1 = y_pred1.rename(columns={0: 'CO2MG811Value'})
```

```python
y_pred1.to_csv('Data_CO2MG811Value.csv')
```
